#Step 11: Handle Missing Values

```python
# Handling numeric missing values

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())
# Handling categorical missing values

df['CouponCode'] = df['CouponCode'].fillna('No Coupon')


```

#Step 12: Check Duplicate Records

```python
df.duplicated().sum()
```

#Step 13 Convert Date Column

```python
df['Date'] = pd.to_datetime(df['Date'], format='mixed', dayfirst=True)
```

#Step 14: Standardize Text Columns

```python
df['Product'] = df['Product'].str.title()
df['OrderStatus'] = df['OrderStatus'].str.title()
```

#Step 15: Dataset Visualization - Histograms

```python
df.hist(figsize=(18,18))
plt.show()
```

#Step 16: Individual Histogram Plots

```python
import random 
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        plt.figure(figsize=(16,8))
        
        c = random.choice(['g','r','b','y','m'])
        
        sns.histplot(df[col], color=c, kde=True)
        
        plt.title(col)
        plt.show()
        
    else:
        print(f"{col} is not numeric")
```

#Step 17: Showing data in PieCharts

```python
df['OrderStatus'].value_counts().plot(
    kind='pie',
    autopct='%.1f%%',
    figsize=(8,8),
    startangle=90
)

plt.title("Order Status Distribution")
plt.ylabel('')
plt.show()
```

```python
df['PaymentMethod'].value_counts().plot(
    kind='pie',
    autopct='%.1f%%',
    figsize=(7,7)
)

plt.title("Payment Method Distribution")
plt.show()
```

```python
df['Product'].value_counts().plot(
    kind='pie',
    autopct='%.1f%%',
    figsize=(8,8)
)

plt.title("Product Distribution")
plt.show()
```

```python
df['ReferralSource'].value_counts().plot(
    kind='pie',
    autopct='%.1f%%',
    figsize=(7,7)
)

plt.title("Referral Source Distribution")
plt.show()
```

```python
df['CouponCode'].value_counts().plot(
    kind='pie',
    autopct='%.1f%%',
    figsize=(7,7)
)

plt.title("Coupon Code Usage")
plt.show()
```
